<h1>1. Bubble Sort </h1>


```python
import array as ar
def bubble_sort(array):
    for i in range(0,len(array)-1):
        for j in range(0,len(array)-i-1):
            if array[j]>array[j+1]:#THIS WILL PRINT IN ASCENDING ORDER 
                array[j],array[j+1]=array[j+1],array[j]
    return array
a=ar.array('i')
try:
    
    size=int(input('enter the size of array '))
    while(len(a)!=size):
        a.append(int(input('enter the value of array ')))

except ValueError:
    print('enter Integer ')
except TypeError:
    print('donot enter null value')

else:
    print('Before sorting ',a)
    print('after sorting in Ascending order',bubble_sort(a))

```

<h1>2.Selection  SOrting</h1>

```python
def insertion_sort(array):
    for i in range(0,len(array)-1):
        min_idx=i
        for j in range(i+1,len(array)):
            if array[min_idx]>array[j]:
                array[min_idx],array[j]=array[j],array[min_idx]
    return array
a=ar.array('i')
try:
    
    size=int(input('enter the size of array '))
    while(len(a)!=size):
        a.append(int(input('enter the value of array ')))

except ValueError:
    print('enter Integer ')
except TypeError:
    print('donot enter null value')

else:
    print('Before sorting ',a)
    print('after sorting in Ascending order',insertion_sort(a))
                
```

## 3.Insertion Sorting

```python
a=[124,56,468,4,67,8,9,322]
for i in range(1,len(a)):
    key=a[i]
    j=i-1
    while j>=0 and a[j]>key:
        a[j+1]=a[j]
        j-=1
    a[j+1]=key
print(a)
```

## 4.Merge Sort  

```python
def merge_sort(array):
    if len(array)>1:
        mid=len(array)//2
        left=array[:mid]
        right=array[mid:]
        merge_sort(left)
        merge_sort(right)
        i=j=k=0
        while i<len(left) and j<len(right):
            
            if left[i]<right[j]:
                array[k]=left[i]
                i+=1
            else:
                array[k]=right[j]
                j+=1
            k=k+1
        while i<len(left):
            array[k]=left[i]
            i+=1
            k+=1
        while j<len(right):
            array[k]=right[j]
            j+=1
            k+=1
a=[122,3,45,6,888,63,114,0,73]
merge_sort(a)
print('sorted array',a)
        
            
```
